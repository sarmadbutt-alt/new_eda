```python
def re_size(size):
    if isinstance(size,str):
        size=size.replace(',','')
        if 'M' in size:
            return float(size.replace('M',''))
        elif 'k'in size:
            return float(size.replace('k',''))
        elif 'Varies with device' in size:
            return np.nan
        return size
```

```python
df['Size']=df['Size'].apply(re_size)
```
