```python
def re_size(size):
    if isinstance(size,str):
        size=size.replace(',','')
        if 'M' in size:
            return float(size.replace('M',''))
        elif 'k'in size:
            return float(size.replace('k',''))
        elif 'G' in size:
            return float(size.replace('G',''))
        elif 'Varies with device' in size:
            return np.nan
        return size
```

```python
df['Size']=df['Size'].apply(re_size)
```

# **Feature Engineering**

```python
bins=[-1,0, 10, 1000, 10000, 100000, 1000000, 10000000, 10000000000]
labels=['no','Very low','low','Moderate','More than Moderate','High','Very High','Top Notch']
df['Installs_Category'] = pd.cut(df['Installs'],bins=bins,labels=labels) 
```

```python
df.head()
```

```python
df.groupby('Installs_Category')['Rating'].describe()
```

```python
df['Installs_Category'].loc[df['Rating'].isnull()].value_counts()
```

```python
# Step 1: group-wise fill
df['Rating'] = df.groupby('Installs_Category')['Rating'].transform(
    lambda x: x.fillna(x.mean())
)

# Step 2: remaining nulls → overall mean
df['Rating'].fillna(df['Rating'].mean(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df[df['Rating Count'].duplicated()].head(10)
```

```python
df['Type'] = df['Free'].map({True: 'Free', False: 'Paid'})
df.head(10)
```

```python
df.drop('Free',axis=1,inplace=True)
```

```python
df.head(10)
```

```python
df['Rating Count'].median()
```

```python
df['Rating Count'].fillna(df['Rating Count'].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Minimum Android']
```

```python
df['Minimum Android'] = df['Minimum Android'].str.extract(r'(\d+\.\d+)')
df['Minimum Android'] = df['Minimum Android'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.dropna(subset='Size',inplace=True)
```
